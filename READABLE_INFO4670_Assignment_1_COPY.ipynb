**Your question:** Does student success correlate with greater advisor attention?



```python
students.info()
```

```python
copy = pd.cut(students["final_gpa"], bins=[0.0, 2.5, 3.5, 4.0],
       labels=["Low Performing", "Mid Performing", "High Performing"])
students_copy = copy.to_frame()
students_copy.head()
```

```python
students_copy = students_copy.join(students["advisor_meetings"].astype(int))
students_copy.head()
```

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---

sns.color_palette("mako", as_cmap=True)
fig, ax = plt.subplots()

sns.boxplot(data=students_copy, x="final_gpa", y="advisor_meetings", hue="final_gpa",
            legend=False, palette="flare", ax=ax)
plt.xticks(rotation=90)
ax.set_title("Advisor Meetings Distribution by GPA Category")
ax.set_xlabel("GPA Performance Category")
ax.set_ylabel("No. of Advisor Meetings")

plt.tight_layout()
plt.show()
```
